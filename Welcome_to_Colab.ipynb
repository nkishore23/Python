```
marks = []
for i in range(5):
  mark = int(input("Enter the mark of Sub" + str(i + 1) + ": "))
  marks.append(mark)


print("=" * 30)
print("Result")
print("=" * 30)

pass_count = 0
fail_count = 0

for m in marks:
  if m >= 35:
    print(m, " Pass")
    pass_count += 1
  else:
    print(m, " Fail")
    fail_count += 1

print("=" * 30)
print("Total Pass: ", pass_count)
print("Total Fail: ", fail_count)
print("Average", sum(marks) / len(marks))
```
